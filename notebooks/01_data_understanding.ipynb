# Phase 2 : Data Cleaning + Exploratory Data Analysis

```python
# Make a working copy
df_clean = df.copy()
```

```python
# Checking column names
df_clean.columns.tolist()
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
)
```

```python
df_clean.columns
```

```python
# Fix TotalCharges
df_clean["totalcharges"].dtype
```

```python
(df_clean["totalcharges"].str.strip() == "").sum()
```

```python
# Let's inspect 
df_clean[
    df_clean["totalcharges"].str.strip() == ""
][["customerid", "tenure", "monthlycharges", "totalcharges"]]
```

```python
# Convert TotalCharges to numeric
df_clean["totalcharges"] = pd.to_numeric(
    df_clean["totalcharges"],
    errors="coerce"
)
```

```python
df_clean["totalcharges"].isnull().sum()
```

```python
df_clean["totalcharges"] = df_clean["totalcharges"].fillna(0)
```

```python
df_clean["totalcharges"].isnull().sum()
```

```python
# check datatypes
df_clean["totalcharges"].dtype
```

```python
# 
df_clean.dropna()
```

```python
# Check duplicates
df_clean.duplicated().sum()
```

```python
df_clean["customerid"].duplicated().sum()
```

```python
# Check categorical values
categorical_cols = df_clean.select_dtypes(
    include="object"
).columns

categorical_cols
```

```python
for col in categorical_cols:
    print(f"\n{col}")
    print(df_clean[col].value_counts())
```

```python
# Convert churn target
df_clean["churn_flag"] = df_clean["churn"].map({
    "Yes": 1,
    "No": 0
})
```
